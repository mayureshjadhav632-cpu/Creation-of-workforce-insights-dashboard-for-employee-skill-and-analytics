SQL Task 6 — Workforce KPI Calculation
Overall KPIs

```python
pd.read_sql_query("""
SELECT
    COUNT(*) AS Total_Employees,
    ROUND(AVG(Salary), 2) AS Average_Salary,
    ROUND(AVG(Performance_Score), 2) AS Average_Performance,
    ROUND(AVG(Engagement_Score), 2) AS Average_Engagement,
    ROUND(AVG(Experience), 2) AS Average_Experience
FROM employees
""", conn)
```

SQL Task 7 — Identify High & Low Performers

```python
pd.read_sql_query("""
SELECT
    Employee_ID,
    Name,
    Department,
    Performance_Score
FROM employees
WHERE Performance_Score >= 4.5
ORDER BY Performance_Score DESC
""", conn)
```

SQL Task 8 — Attrition Analysis

```python
pd.read_sql_query("""
SELECT
    Attrition,
    COUNT(*) AS Employee_Count
FROM employees
GROUP BY Attrition
""", conn)
```

SQL Task 9 — Department-wise Attrition & Retention

```python
pd.read_sql_query("""
SELECT
    Department,
    COUNT(*) AS Total_Employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS Attrition_Count,
    SUM(CASE WHEN Attrition = 'No' THEN 1 ELSE 0 END) AS Retained_Count,
    ROUND(
        100.0 * SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END)
        / COUNT(*),
        2
    ) AS Attrition_Rate
FROM employees
GROUP BY Department
ORDER BY Attrition_Rate DESC
""", conn)
```

SQL Task 10 — Employee Performance Analysis

```python
pd.read_sql_query("""
SELECT
    Department,
    AVG(Performance_Score) AS Average_Performance
FROM employees
GROUP BY Department
ORDER BY Average_Performance DESC;
""", conn)
```

```python
pd.read_sql_query("""
SELECT
    Department,
    COUNT(*) AS Total_Employees,
    ROUND(AVG(Performance_Score), 2) AS Average_Performance,
    MAX(Performance_Score) AS Highest_Performance,
    MIN(Performance_Score) AS Lowest_Performance
FROM employees
GROUP BY Department
ORDER BY Average_Performance DESC;
""", conn)
```
