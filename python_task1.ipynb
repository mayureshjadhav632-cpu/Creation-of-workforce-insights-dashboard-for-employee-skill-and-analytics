## Task 4: Workforce KPI Analysis

Calculate important workforce Key Performance Indicators (KPIs) from the employee dataset, including total employees, average age, average salary, average engagement score, average job satisfaction, and employee attrition rate.

```python
# Calculate Workforce KPIs

total_employees = df["Employee_ID"].count()
average_age = df["Age"].mean()
average_salary = df["Salary"].mean()
average_engagement = df["Engagement_Score"].mean()
average_satisfaction = df["Job_Satisfaction"].mean()

attrition_rate = (df["Attrition"].value_counts(normalize=True)["Yes"]) * 100

print("Workforce KPIs")
print("-----------------------------")
print("Total Employees:", total_employees)
print("Average Age:", round(average_age, 2))
print("Average Salary:", round(average_salary, 2))
print("Average Engagement Score:", round(average_engagement, 2))
print("Average Job Satisfaction:", round(average_satisfaction, 2))
print("Attrition Rate:", round(attrition_rate, 2), "%")
```

## Task 5: Department-wise Workforce Analysis

Analyze the distribution of employees across different departments. Calculate the number of employees in each department and visualize the department-wise workforce distribution using a bar chart.

```python
# Calculate employee count by department

department_count = df["Department"].value_counts()

print("Employees by Department:")
print(department_count)
```

```python
# Visualize employee distribution by department

fig = px.bar(
    department_count,
    x=department_count.index,
    y=department_count.values,
    labels={
        "x": "Department",
        "y": "Number of Employees"
    },
    title="Employee Distribution by Department"
)

fig.show()
```

## Task 6: Salary Analysis

Analyze employee salary information using Python. Calculate the average salary, minimum salary, and maximum salary, and determine the average salary for each department. Visualize the average salary across departments to understand salary distribution within the workforce.

```python
# Calculate overall salary statistics

average_salary = df["Salary"].mean()
minimum_salary = df["Salary"].min()
maximum_salary = df["Salary"].max()

print("Salary Analysis")
print("-----------------------------")
print("Average Salary:", round(average_salary, 2))
print("Minimum Salary:", minimum_salary)
print("Maximum Salary:", maximum_salary)
```

```python
# Calculate average salary by department

department_salary = df.groupby("Department", as_index=False)["Salary"].mean()

print("Average Salary by Department:")
print(department_salary)
```

```python
# Visualize average salary by department

fig = px.bar(
    department_salary,
    x="Department",
    y="Salary",
    title="Average Salary by Department",
    labels={
        "Department": "Department",
        "Salary": "Average Salary"
    }
)

fig.show()
```

## Task 7: Employee Engagement and Job Satisfaction Analysis

Analyze employee engagement and job satisfaction scores to understand overall workforce experience and identify differences between departments.

```python
# Calculate average engagement and satisfaction

avg_engagement = df["Engagement_Score"].mean()
avg_satisfaction = df["Job_Satisfaction"].mean()

print("Average Engagement Score:", round(avg_engagement, 2))
print("Average Job Satisfaction:", round(avg_satisfaction, 2))
```

Analyze employee attrition using Python by calculating the number and percentage of employees who have left the organization.

## Task 9: Workforce Performance Visualization

Visualize employee performance scores across departments to identify differences in workforce performance.

## Task 10: Employee Segmentation using K-Means

Apply K-Means Clustering to group employees based on workforce characteristics such as salary, experience, performance, engagement, and job satisfaction.

```python
# Select features for employee segmentation

features = [
    "Age",
    "Experience",
    "Salary",
    "Performance_Score",
    "Job_Satisfaction",
    "Engagement_Score"
]

X = df[features]

# Standardize the features

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply K-Means clustering

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

df["Employee_Cluster"] = kmeans.fit_predict(X_scaled)

print("Employee Segments:")
print(df[["Employee_ID", "Employee_Cluster"]])
```

```python
# Visualize employee clusters

fig = px.scatter(
    df,
    x="Experience",
    y="Salary",
    color="Employee_Cluster",
    title="Employee Segmentation using K-Means",
    hover_data=["Employee_ID", "Department"]
)

fig.show()
```
